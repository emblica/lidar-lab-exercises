# Task 4: Static vs dynamic: Movement detector
* **Goal:** Find and visualize the static surfaces and moving objects
* **Why:** A critical pre-processing step for many applications, like separating static (ground/walls) from dynamic (people) objects.

* **Ideas:** 
    * Store the first frame's (downsampled) point cloud as a "background". Compare it to the current frame. Color any points in the current frame that don't have a close neighbor in the previous frame.
    * Advanced version: Update the background on the go
    * Try different threshold for defining the "close neighbour"
    * Search for libraries for nearest neighbour calculation (like `KDTree` from `scipy.spatial`)
    * Alternative solution: Voxel grid background:
        * Divide the 3D space into small cubes (voxels). Instead of storing the bacground points, you store an "occupied/not occupied" label for each voxel
