# Tier 0: Getting started

> Recommended: set up an isolated Python environment so everyone has the same libraries.

## Environments
This project targets Python 3.11 or 3.12.

Pick ONE of the options below. The options are in the recommendation order.

### Option A — uv (fastest, recommended)
uv can create a local `.venv` and install dependencies straight from `pyproject.toml`. Install from here: https://docs.astral.sh/uv/getting-started/installation/

1) Ensure uv is installed
   - Check: `uv version`
2) From the repository root, install deps:
```bash
uv sync
```

You can then run scripts via:
```bash
uv run <your_script>.py
```

You can test if the environment and libraries are working by running:
```bash
uv run scripts/helloworld.py
```

Add new libraries like this, if needed:
```bash
uv add <your-library>
```

#### Activate environment in notebook:
To run this notebook with the same environment, register a Jupyter kernel from your venv.

Not tested options, but should work as well:

### Option B — pyenv + pip
Use pyenv to get an exact Python version, then install dependencies.

```bash
# Create virtualenv
pyenv virtualenv 3.12.2 <your-env-name>
# Activate
pyenv activate <your-env-name>
# Install deps
pip install -r requirements.txt
```
Run scripts:
```bash
python your-script.py
```

### Option C — venv + pip 
Use your system Python 3.12+ directly.

```bash
python -m venv .venv
source .venv/bin/activate
pip install -r requirements.txt
```

### Option D — Global install (not recommended)
If you must, install packages into your user site-packages (avoids sudo):
```bash
pip install -r requirements.txt
```
Be aware this may conflict with other projects on your machine.
